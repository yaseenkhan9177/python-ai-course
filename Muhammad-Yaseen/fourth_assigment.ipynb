```python
import pandas as pd
df = pd.read_csv(r"E:\data sets\loan_data_set.csv")
```

```python
missing_flag = df.isnull().any()
print("Columns with missing values:")
print(missing_flag[missing_flag].index.tolist())
```

```python
missing_count = df.isnull().sum()
non_missing_count = df.notnull().sum()
missing_pct = (missing_count / len(df) * 100).round(2)

missing_stats = pd.DataFrame({
    "Missing_Values": missing_count,
    "Non_Missing_Values": non_missing_count,
    "Missing_Percentage": missing_pct
}).sort_values("Missing_Values", ascending=False)

missing_stats
```

```python
total_missing = df.isnull().sum().sum()
total_non_missing = df.notnull().sum().sum()
total_cells = df.shape[0] * df.shape[1]
overall_missing_pct = round(total_missing / total_cells * 100, 2)

print(f"Total missing values in dataset : {total_missing}")
print(f"Total non-missing values        : {total_non_missing}")
print(f"Total cells (rows x columns)     : {total_cells}")
print(f"Overall percentage of empty cells: {overall_missing_pct}%")
```

```python
problematic = missing_stats[missing_stats["Missing_Values"] > 0].sort_values("Missing_Values", ascending=False)
problematic
```

```python
top_var = problematic.index[0]
print(f"Most problematic variable: '{top_var}' "
      f"with {problematic.loc[top_var, 'Missing_Values']} missing values "
      f"({problematic.loc[top_var, 'Missing_Percentage']}% of records).")
```

```python
plt.figure(figsize=(10, 6))
ax = sns.barplot(
    x=problematic["Missing_Values"],
    y=problematic.index,
    hue=problematic.index,
    palette="Reds_r",
    legend=False
)
ax.set_title("Missing Values by Column — Loan Dataset", fontsize=14, fontweight="bold")
ax.set_xlabel("Number of Missing Values")
ax.set_ylabel("Column")

for i, v in enumerate(problematic["Missing_Values"]):
    ax.text(v + 0.5, i, str(v), va="center", fontsize=10)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap="viridis", yticklabels=False)
plt.title("Missing Data Heatmap — Loan Dataset", fontsize=14, fontweight="bold")
plt.xlabel("Columns")
plt.tight_layout()
plt.show()
```

```python
student_data = {
    "Student": ["Ayesha", "Bilal", "Fatima", "Hassan", "Sara"],
    "Math": [88, 72, 95, 60, 79],
    "Science": [91, 68, 89, 74, 85],
    "English": [76, 81, 92, 65, 70]
}

students_df = pd.DataFrame(student_data)
students_df["Average"] = students_df[["Math", "Science", "English"]].mean(axis=1).round(2)
students_df
```

```python
plt.figure(figsize=(8, 5))
ax = sns.barplot(
    data=students_df,
    x="Student", y="Average",
    hue="Student",
    palette="viridis",
    legend=False
)
ax.set_title("Average Marks by Student", fontsize=14, fontweight="bold")
ax.set_ylabel("Average Marks (out of 100)")
ax.set_ylim(0, 100)

for i, v in enumerate(students_df["Average"]):
    ax.text(i, v + 1, str(v), ha="center", fontsize=10)

plt.tight_layout()
plt.show()
```

```python
students_long = students_df.melt(
    id_vars="Student",
    value_vars=["Math", "Science", "English"],
    var_name="Subject",
    value_name="Marks"
)

plt.figure(figsize=(9, 5))
sns.barplot(data=students_long, x="Student", y="Marks", hue="Subject", palette="Set2")
plt.title("Marks by Subject for Each Student", fontsize=14, fontweight="bold")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
ax = sns.countplot(
    data=df,
    x="Property_Area",
    hue="Property_Area",
    order=df["Property_Area"].value_counts().index,
    palette="pastel",
    legend=False
)
ax.set_title("Applicant Count by Property Area", fontsize=14, fontweight="bold")
ax.set_xlabel("Property Area")
ax.set_ylabel("Number of Applicants")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```
