```python
import matplotlib.pyplot as plt

students = ["Ali", "Ahmad", "Hamza", "Bilal", "Usman"]
marks = [78, 65, 90, 55, 82]

plt.plot(students, marks, label="Student Marks")

plt.title("Student Performance")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib import style

style.use("ggplot")

x = [1, 5, 4]
y = [23, 45, 89]

x2 = [3, 7, 3]
y2 = [5, 8, 9]

plt.plot(x, y, linewidth=3, label="first")
plt.plot(x2, y2, linewidth=3, label="second")

plt.title("Comparison of Two Lines")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.legend()

plt.show()
```
