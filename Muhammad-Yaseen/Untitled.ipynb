```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt



data = np.array([
    [1, 2, 3, 4],
    [5, 4, 3, 4],
    [2, 3, 4, 5],
    [3, 1, 2, 5],
    [2, 1, 5, 7]
])

ans = pd.DataFrame(data)
print(ans)
```

```python
print(ans.head(3))


print(ans.head(2))

print(ans.tail(2))
```

```python
print(ans.tail(3))
```

```python
row1 = (ans.iloc[2])
print(f"this is thir row in from data frame  \n{row1}")
row2=(ans.iloc[0:2])
print(f"this is [0:2]  is \n{row2}")       
```

```python
columns =(ans.iloc[: , 0:2])
print(columns)
```

```python
columnsrows = (ans.iloc[0:3, 0:2])
print(columnsrows)
```
