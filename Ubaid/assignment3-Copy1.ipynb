```python

from matplotlib import pyplot as plt
from matplotlib import style
x = [1,5,4]
y = [23,45,89]
plt.title('Linear Equition Model')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.plot(x,y)
plt.show()
```

```python
style.use("grayscale")
x = [1, 5, 4]
y = [23, 45, 89]


x2 = [3, 7, 3]
y2 = [5, 8, 9]

plt.plot(x,y)
plt.plot(x2,y2)
plt.plot(x,y, label='first',linewidth=5)
plt.plot(x2,y2, label='second')
plt.legend()
plt.title(" graph with style")
plt.show()
```

```python
x = [1,2,4,5,7,9]
y = [5,4,10,2,8,6]
plt.bar(x,y)
plt.show()
plt.hist(y)
```

```python
x = [1, 5, 4, 7]
y = [5, 4, 10, 8]
plt.scatter(x,y)
plt.title("scatter")
plt.show()
```

```python
students = ["ali","ahmad","hamza","bilal","usman"]
marks = [10,20,50,80,90]
plt.plot(students,marks)
plt.xlabel("students")
plt.ylabel("marks")
plt.bar(students,marks)
plt.scatter(students,marks)
plt.title("students result")

plt.show
```

```python
plt.hist(marks)
```

```python
style.use("dark_background")
x = [1, 5, 4]
y = [23, 45, 89]


x2 = [3, 7, 3]
y2 = [5, 8, 9]

plt.plot(x,y, label="line 1",linewidth=3,color="blue")
plt.plot(x2,y2, label="line 2",linewidth=3,color="red")
plt.title("two line graph")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend()
```
